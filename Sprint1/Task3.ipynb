```
using System;

public class ShippingOption
{
  
    private int _deliveryOptionId;
    private string _deliveryOptionName;
    private double _cost;


    public int DeliveryOptionId
    {
        get { return _deliveryOptionId; }
        set { _deliveryOptionId = value; }
    }

    public string DeliveryOptionName
    {
        get { return _deliveryOptionName; }
        set { _deliveryOptionName = value; }
    }

    public double Cost
    {
        get { return _cost; }
        set
        {
            if (value >= 0)
                _cost = value;
            else
                Console.WriteLine("Стоимость не может быть отрицательной!");
        }
    }


    public ShippingOption(int id, string name, double cost)
    {
        DeliveryOptionId = id;
        DeliveryOptionName = name;
        Cost = cost;
    }

   
    public virtual double CalculateCost()
    {
        return Cost;
    }

    public virtual int EstimateDeliveryTime()
    {
        return 5;
    }

    public virtual string GetDeliveryDetails()
    {
        return $"Опция №{DeliveryOptionId}: {DeliveryOptionName}, " +
               $"стоимость: {CalculateCost()} руб., срок: {EstimateDeliveryTime()} дн.";
    }
}
```

```
public class StandardDelivery : ShippingOption
{
    public int AverageDeliveryTime { get; set; }

    public StandardDelivery(int id, string name, double cost, int averageDeliveryTime)
        : base(id, name, cost)
    {
        AverageDeliveryTime = averageDeliveryTime;
    }

    public override int EstimateDeliveryTime()
    {
        return AverageDeliveryTime;
    }
}

public class ExpressDelivery : ShippingOption
{
    public int MinDeliveryTime { get; set; }
    public double UrgencySurcharge { get; set; }

    public ExpressDelivery(int id, string name, double cost, int minDeliveryTime, double urgencySurcharge)
        : base(id, name, cost)
    {
        MinDeliveryTime = minDeliveryTime;
        UrgencySurcharge = urgencySurcharge;
    }

    public override double CalculateCost()
    {
        return Cost + UrgencySurcharge;
    }

    public override int EstimateDeliveryTime()
    {
        return MinDeliveryTime;
    }
}

public class Pickup : ShippingOption
{
    public string PickupAddress { get; set; }

    public Pickup(int id, string name, double cost, string pickupAddress)
        : base(id, name, cost)
    {
        PickupAddress = pickupAddress;
    }

    public override double CalculateCost()
    {
        return 0;
    }

    public override string GetDeliveryDetails()
    {
        return $"Опция №{DeliveryOptionId}: {DeliveryOptionName}. " +
               $"Адрес пункта самовывоза: {PickupAddress}. Стоимость: {CalculateCost()} руб.";
    }
}
```

```
public class ShippingManager
{
    public string ManagerName { get; set; }

    public ShippingManager(string name)
    {
        ManagerName = name;
    }

    public void ShowOption(ShippingOption option)
    {
        Console.WriteLine($"Менеджер {ManagerName} рассматривает опцию:");
        Console.WriteLine(option.GetDeliveryDetails());

    }

    public void CompareCost(ShippingOption first, ShippingOption second)
    {
        double cost1 = first.CalculateCost();
        double cost2 = second.CalculateCost();

        Console.WriteLine($"{ManagerName} сравнивает опции:");
        Console.WriteLine($"{first.DeliveryOptionName}: {cost1} руб.");
        Console.WriteLine($"{second.DeliveryOptionName}: {cost2} руб.");

        if (cost1 < cost2)
            Console.WriteLine($"Дешевле: {first.DeliveryOptionName}");
        else if (cost2 < cost1)
            Console.WriteLine($"Дешевле: {second.DeliveryOptionName}");
        else
            Console.WriteLine("Стоимость одинаковая.");

    }
}
```

```
StandardDelivery standard = new StandardDelivery(1, "Стандартная доставка", 500, 7);
ExpressDelivery express = new ExpressDelivery(2, "Экспресс-доставка", 500, 2, 500);
Pickup pickup = new Pickup(3, "Самовывоз", 0, "г. Тюмень, ул. Мельникайте, д. 142");

ShippingManager manager = new ShippingManager("Лера");

manager.ShowOption(standard);
manager.ShowOption(express);
manager.ShowOption(pickup);
manager.CompareCost(standard, express);
manager.CompareCost(express, pickup);
```
