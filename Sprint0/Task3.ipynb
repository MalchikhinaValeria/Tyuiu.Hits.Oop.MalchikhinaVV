### Вариант задания №17 


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс ShippingOption в C#, который будет представлять
различные опции доставки. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

<h3><b>Требования к базовому классу ShippingOption:</b></h3>
<h4>• Атрибуты:</h4> ID опции доставки (DeliveryOptionId),<br> Название опции доставки
(DeliveryOptionName),<br> Стоимость доставки (Cost).<br>
<h4>• Методы:</h4>
o CalculateCost(): метод для расчета стоимости доставки.<br>
o EstimateDeliveryTime(): метод для оценки времени доставки.<br>
o GetDeliveryDetails(): метод для получения деталей опции доставки.<br>
<h3><b>Требования к производным классам:</b></h3>
<h4>1. СтандартнаяДоставка (StandardDelivery):</h4> Должна содержать
дополнительные атрибуты, такие как Среднее время доставки
(AverageDeliveryTime).<br> Метод EstimateDeliveryTime() должен быть
переопределен для предоставления среднего времени доставки.<br>
<h4>2. ЭкспрессДоставка (ExpressDelivery):</h4> Должна содержать дополнительные
атрибуты, такие как Минимальное время доставки (MinDeliveryTime).<br>
Метод CalculateCost() должен быть переопределен для увеличения
стоимости доставки в случае необходимости ускорения доставки.<br>
<h4>3. Самовывоз (Pickup) (если требуется третий класс):</h4> Должна содержать
дополнительные атрибуты, такие как Адрес пункта самовывоза
(PickupAddress). <br>Метод GetDeliveryDetails() должен быть переопределен для
отображения адреса пункта самовывоза.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class ShippingOption {


    public int DeliveryOptionId { get; set; }
    
    public string DeliveryOptionName { get; set; }

    public double Cost { get; set; }

    public ShippingOption(int id, string name, double cost)
    {
        DeliveryOptionId = id;
        DeliveryOptionName = name;
        Cost = cost;
    }

     public virtual double CalculateCost()
    {
        return Cost;
    }

    public virtual int EstimateDeliveryTime()
    {
        return 7;
    }

    public virtual string GetDeliveryDetails()
    {
        return $"Опция №{DeliveryOptionId}: {DeliveryOptionName}, " +
               $"стоимость: {CalculateCost()} руб., срок: {EstimateDeliveryTime()} дн.";
    }
 
}
```

```
public class StandardDelivery : ShippingOption
{

    public int AverageDeliveryTime { get; set; }

    public StandardDelivery(int id, string name, double cost, int averageDeliveryTime)
        : base(id, name, cost) 
    {
        AverageDeliveryTime = averageDeliveryTime;
    }

    public override int EstimateDeliveryTime()
    {
        return AverageDeliveryTime;
    }
}


public class ExpressDelivery : ShippingOption
{

    public int MinDeliveryTime { get; set; }
    public double ExtraCharge { get; set; } 

    public ExpressDelivery(int id, string name, double cost, int minDeliveryTime, double extraCharge)
        : base(id, name, cost)
    {
        MinDeliveryTime = minDeliveryTime;
        ExtraCharge = extraCharge;
    }


    public override double CalculateCost()
    {
        return Cost + ExtraCharge;
    }

 
    public override int EstimateDeliveryTime()
    {
        return MinDeliveryTime;
    }
}


public class Pickup : ShippingOption
{

    public string PickupAddress { get; set; }

    public Pickup(int id, string name, double cost, string pickupAddress)
        : base(id, name, cost)
    {
        PickupAddress = pickupAddress;
    }


    public override double CalculateCost()
    {
        return 0;
    }


    public override string GetDeliveryDetails()
    {
        return $"Опция №{DeliveryOptionId}: {DeliveryOptionName}. " +
               $"Адрес пункта самовывоза: {PickupAddress}. Стоимость: {CalculateCost()} руб.";
    }
}
```

```
StandardDelivery standard = new StandardDelivery(1, "Стандартная доставка", 500, 7);
ExpressDelivery express = new ExpressDelivery(2, "Экспресс-доставка", 500, 2, 500);
Pickup pickup = new Pickup(3, "Самовывоз", 0, "г. Тюмень, ул. Мельникайте, д. 142");

Console.WriteLine(standard.GetDeliveryDetails());

Console.WriteLine(express.GetDeliveryDetails());

Console.WriteLine(pickup.GetDeliveryDetails());
```
