<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс ShippingOption в C#, который будет представлять
различные опции доставки. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

public class ShippingOption
{

    private int deliveryOptionId;
    private string deliveryOptionName;
    private double cost;
    protected string carrier;  
    internal string region;      

    public int DeliveryOptionId
    {
        get { return deliveryOptionId; }
        set { deliveryOptionId = value; }
    }

    public string DeliveryOptionName
    {
        get { return deliveryOptionName; }
        set { deliveryOptionName = value; }
    }

    public double Cost
    {
        get { return cost; }
        set
        {
            if (value >= 0)
                cost = value;
            else
                Console.WriteLine("Стоимость не может быть отрицательной!");
        }
    }

    public string Carrier
    {
        get { return carrier; }
        set { carrier = value; }
    }

    public string Region
    {
        get { return region; }
        set { region = value; }
    }


    public ShippingOption(int id, string name, double cost, string carrier, string region)
    {
        DeliveryOptionId = id;
        DeliveryOptionName = name;
        Cost = cost;
        Carrier = carrier;
        Region = region;
    }


    public virtual double CalculateCost()
    {
        return Cost;
    }

    public virtual int EstimateDeliveryTime()
    {
        return 5;
    }

    public virtual string GetDeliveryDetails()
    {
        return $"Опция №{DeliveryOptionId}: {DeliveryOptionName} | Перевозчик: {Carrier} | " +
               $"Регион: {Region} | Стоимость: {CalculateCost()} руб. | Срок: {EstimateDeliveryTime()} дн.";
    }


    public virtual bool IsAvailableInRegion(string targetRegion)
    {
        return Region == targetRegion;
    }
}

public class StandardDelivery : ShippingOption
{
    public int AverageDeliveryTime { get; set; }
    public bool IsTrackable { get; set; }

    public StandardDelivery(int id, string name, double cost, string carrier, string region,
                            int averageDeliveryTime, bool isTrackable)
        : base(id, name, cost, carrier, region)
    {
        AverageDeliveryTime = averageDeliveryTime;
        IsTrackable = isTrackable;
    }

    public override int EstimateDeliveryTime()
    {
        return AverageDeliveryTime;
    }


    public void NotifyCustomer()
    {
        Console.WriteLine($"Уведомление: стандартная доставка «{DeliveryOptionName}» принята.");
    }
}


public class ExpressDelivery : ShippingOption
{
    public int MinDeliveryTime { get; set; }
    public double UrgencySurcharge { get; set; }
    public bool HasInsurance { get; set; }

    public ExpressDelivery(int id, string name, double cost, string carrier, string region,
                           int minDeliveryTime, double urgencySurcharge, bool hasInsurance)
        : base(id, name, cost, carrier, region)
    {
        MinDeliveryTime = minDeliveryTime;
        UrgencySurcharge = urgencySurcharge;
        HasInsurance = hasInsurance;
    }

    public override double CalculateCost()
    {
        return Cost + UrgencySurcharge;
    }

    public override int EstimateDeliveryTime()
    {
        return MinDeliveryTime;
    }

    public void NotifyUrgent()
    {
        Console.WriteLine($"Срочно! Экспресс-доставка «{DeliveryOptionName}» оформлена.");
    }
}


public class Pickup : ShippingOption
{
    public string PickupAddress { get; set; }
    public string WorkingHours { get; set; }
    public bool NeedsID { get; set; }

    public Pickup(int id, string name, double cost, string carrier, string region,
                  string pickupAddress, string workingHours, bool needsID)
        : base(id, name, cost, carrier, region)
    {
        PickupAddress = pickupAddress;
        WorkingHours = workingHours;
        NeedsID = needsID;
    }

    public override double CalculateCost()
    {
        return 0;
    }

    public override string GetDeliveryDetails()
    {
        return $"Опция №{DeliveryOptionId}: {DeliveryOptionName} | Адрес: {PickupAddress} | " +
               $"Часы работы: {WorkingHours} | Нужен ID: {(NeedsID ? "да" : "нет")} | " +
               $"Стоимость: {CalculateCost()} руб.";
    }

    public void ShowMap()
    {
        Console.WriteLine($"Пункт самовывоза находится по адресу: {PickupAddress}");
    }
}

public class SameDayDelivery : ExpressDelivery
{
    public string CutoffTime { get; set; }      
    public bool OnlyWithinCity { get; set; }
    public double ExtraFee { get; set; }

    public SameDayDelivery(int id, string name, double cost, string carrier, string region,
                           int minDeliveryTime, double urgencySurcharge, bool hasInsurance,
                           string cutoffTime, bool onlyWithinCity, double extraFee)
        : base(id, name, cost, carrier, region, minDeliveryTime, urgencySurcharge, hasInsurance)
    {
        CutoffTime = cutoffTime;
        OnlyWithinCity = onlyWithinCity;
        ExtraFee = extraFee;
    }

    public override double CalculateCost()
    {
        return base.CalculateCost() + ExtraFee;
    }

    public override int EstimateDeliveryTime()
    {
        return 0; 
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() +
               $" | Только по городу: {(OnlyWithinCity ? "да" : "нет")} | Приём заказов до: {CutoffTime}";
    }

    public void NotifySameDay()
    {
        Console.WriteLine($"Доставка в тот же день! Заказы принимаются до {CutoffTime}.");
    }
}

public interface ITrackable
{
    void Track();
}


public interface IInsurable
{
    void Insure(double amount);
}

public class InternationalDelivery : ShippingOption, ITrackable, IInsurable
{
    public string Country { get; set; }
    public double CustomsFee { get; set; }
    public double InsuranceAmount { get; set; }

    public InternationalDelivery(int id, string name, double cost, string carrier, string region,
                                 string country, double customsFee)
        : base(id, name, cost, carrier, region)
    {
        Country = country;
        CustomsFee = customsFee;
        InsuranceAmount = 0;
    }

    public override double CalculateCost()
    {
        return Cost + CustomsFee + InsuranceAmount;
    }

    public override int EstimateDeliveryTime()
    {
        return 14;
    }

    public override string GetDeliveryDetails()
    {
        return base.GetDeliveryDetails() +
               $" | Страна: {Country} | Таможенный сбор: {CustomsFee} руб.";
    }


    public void Track()
    {
        Console.WriteLine($"Международная посылка «{DeliveryOptionName}» отслеживается. Страна: {Country}");
    }


    public void Insure(double amount)
    {
        InsuranceAmount = amount;
        Console.WriteLine($"Посылка застрахована на сумму {amount} руб.");
    }
}

StandardDelivery standard = new StandardDelivery(1, "Стандартная доставка", 300, "Почта России", "Тюмень", 7, true);
ExpressDelivery express = new ExpressDelivery(2, "Экспресс-доставка", 300, "СДЭК", "Тюмень", 2, 500, true);
Pickup pickup = new Pickup(3, "Самовывоз", 0, "Магазин", "Тюмень", "ул. Республики, 142", "09:00-21:00", false);


SameDayDelivery sameDay = new SameDayDelivery(4, "Доставка в тот же день", 400, "Курьер", "Тюмень", 0, 700, true, "14:00", true, 200);

InternationalDelivery international = new InternationalDelivery(5, "Международная доставка", 1500,"DHL", "Москва", "Германия", 800);


Console.WriteLine(standard.GetDeliveryDetails());
Console.WriteLine(express.GetDeliveryDetails());
Console.WriteLine(pickup.GetDeliveryDetails());
standard.NotifyCustomer();
express.NotifyUrgent();
pickup.ShowMap();


Console.WriteLine(sameDay.GetDeliveryDetails());
sameDay.NotifySameDay();


Console.WriteLine(international.GetDeliveryDetails());
international.Insure(5000);
international.Track();
Console.WriteLine($"Итоговая стоимость: {international.CalculateCost()} руб.");
```
